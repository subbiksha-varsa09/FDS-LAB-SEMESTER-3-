```python
import numpy as np
import pandas as pd

# 1. Creating a DataFrame using a list
list1 = [[1, 'Smith', 50000],
         [2, 'Jones', 60000]]

df1 = pd.DataFrame(list1)
print(df1)

# 2. Giving column names
df1.columns = ['EmpId', 'Name', 'Salary']
print(df1)

# 3. Display DataFrame information
df1.info()

# 4. Reading a CSV file
df = pd.read_csv('employee.csv')

# 5. Display first 5 rows
print(df.head())

# 6. Display last 5 rows
print(df.tail())

# 7. Display DataFrame information
df.info()

# 8. Accessing the salary column
print(df['Salary'])

# 9. Checking the data type
print(type(df['Salary']))

# 10. Calculating statistical values
print("Mean:", df['Salary'].mean())
print("Median:", df['Salary'].median())
print("Mode:", df['Salary'].mode())
print("Variance:", df['Salary'].var())
print("Standard Deviation:", df['Salary'].std())

# 11. Statistical summary
print(df.describe())

# 12. Summary including all columns
print(df.describe(include='all'))

# 13. Display column names
empCol = df.columns
print(empCol)

# 14. Display DataFrame values
emparray = df.values
print(emparray)
```
