```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
url = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"
try:
    diabetes_df = pd.read_csv(url)
    diabetes_df.to_csv("diabetes.csv", index=False)
    print("Diabetes CSV file downloaded and created successfully!")
except Exception as e:
    print("Error downloading the dataset:", e)
    raise
print("\nFirst five rows:")
print(diabetes_df.head())
print("\nLast five rows:")
print(diabetes_df.tail())
print("\nShape of the dataset:")
print(diabetes_df.shape)
print("\nColumn names:")
print(diabetes_df.columns)
print("\nDataset Information:")
diabetes_df.info()
print("\nMissing Values:")
print(diabetes_df.isnull().sum())
print("\nStatistical Summary:")
print(diabetes_df.describe())
print("\nAll columns:")
print(diabetes_df.columns.tolist())
diabetes_df.hist(figsize=(12, 10), bins=20)
plt.suptitle("Diabetes Dataset - Histograms")
plt.tight_layout()
plt.show()
plt.figure(figsize=(10, 8))
sns.heatmap(
    diabetes_df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Diabetes Dataset - Correlation Heatmap")
plt.tight_layout()
plt.show()
print("\nDataset loaded successfully!")
print("Number of rows:", diabetes_df.shape[0])
print("Number of columns:", diabetes_df.shape[1])
print("CSV file saved as diabetes.csv")
```
