````python
import numpy as np
import pandas as pd
data = {
    "Country": [
        "France", "Spain", "Germany", "Spain", "Germany",
        "France", "Spain", np.nan, "France", "France"
    ],
    "Age": [44, 27, 30, 38, 40, 35, np.nan, 48, 50, 37],
    "Salary": [
        72000, 48000, 54000, 61000, np.nan,
        58000, 52000, 79000, 83000, 67000
    ],
    "Purchased": [
        "No", "Yes", "No", "No", "Yes",
        "Yes", "No", "Yes", "No", "Yes"
    ]
}

df = pd.DataFrame(data)
df.to_csv("pre-process_datasample.csv", index=False)
print("Original Dataset:")
print(df)
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))
print("\nDataset After Handling Missing Values:")
print(df)
country_dummies = pd.get_dummies(df["Country"])

# Step 10: Create the updated dataset
updated_dataset = pd.concat(
    [country_dummies, df[["Age", "Salary", "Purchased"]]],
    axis=1
)

print("\nDataset After Encoding Country:")
print(updated_dataset)

# Step 11: Convert Purchased values into numbers
updated_dataset["Purchased"] = updated_dataset["Purchased"].map({
    "No": 0,
    "Yes": 1
})

# Step 12: Display the final dataset
print("\nFinal Preprocessed Dataset:")
print(updated_dataset)

# Step 13: Check missing values
print("\nMissing Values in Final Dataset:")
print(updated_dataset.isnull().sum())

# Step 14: Display final dataset information
print("\nFinal Dataset Information:")
updated_dataset.info()

# Step 15: Save the final dataset
updated_dataset.to_csv("updated_dataset.csv", index=False)

print("\nData preprocessing completed successfully!")
print("Original CSV: pre-process_datasample.csv")
print("Final CSV: updated_dataset.csv")
```
````
