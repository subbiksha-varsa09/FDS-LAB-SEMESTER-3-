```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import gutenberg

# Download required datasets
nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')

# Load the book
sample = gutenberg.raw("austen-emma.txt")

# Tokenize the text
token = word_tokenize(sample)

# Store the first 50 tokens
wlist = []

for i in range(50):
    wlist.append(token[i])

# Count frequency of each unique token
wordfreq = [(w, wlist.count(w)) for w in dict.fromkeys(wlist)]

# Print word-frequency pairs
print("Word Frequency Pairs:")
print(wordfreq)
```
