```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)
data.rename(columns={
    "sepal_length": "SepalLengthCm",
    "sepal_width": "SepalWidthCm",
    "petal_length": "PetalLengthCm",
    "petal_width": "PetalWidthCm",
    "species": "variety"
}, inplace=True)
data.insert(0, "Id", range(1, len(data) + 1))
data.to_csv("Iris_Dataset.csv", index=False)
print("Complete Dataset:")
display(data)
print("\nDataset Information:")
data.info()
print("\nStatistical Description:")
print(data.describe())
print("\nVariety Count:")
print(data.value_counts("variety"))
plt.figure(figsize=(8, 5))
sns.countplot(x="variety", data=data)
plt.title("Count of Iris Varieties")
plt.show()
dummies = pd.get_dummies(data["variety"])
FinalDataset = pd.concat(
    [dummies, data.iloc[:, [0, 1, 2, 3]]],
    axis=1
)
print("\nFinal Dataset:")
display(FinalDataset.head())
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="variety",
    data=data
)
plt.title("Sepal Length vs Sepal Width")
plt.show()
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="variety",
    data=data
)
plt.title("Petal Length vs Petal Width")
plt.show()
sns.pairplot(data, hue="variety", height=3)
plt.show()
sns.FacetGrid(
    data,
    hue="variety",
    height=5
).map(
    sns.histplot,
    "PetalLengthCm"
).add_legend()
plt.title("Petal Length Distribution")
plt.show()
sns.FacetGrid(
    data,
    hue="variety",
    height=5
).map(
    sns.histplot,
    "PetalWidthCm"
).add_legend()
plt.title("Petal Width Distribution")
plt.show()
sns.FacetGrid(
    data,
    hue="variety",
    height=5
).map(
    sns.histplot,
    "SepalLengthCm"
).add_legend()
plt.title("Sepal Length Distribution")
plt.show()
sns.FacetGrid(
    data,
    hue="variety",
    height=5
).map(
    sns.histplot,
    "SepalWidthCm"
).add_legend()
plt.title("Sepal Width Distribution")
plt.show()
print("\nAll operations completed successfully!")
print("Dataset shape:", data.shape)
print("CSV file saved as Iris_Dataset.csv")
```
